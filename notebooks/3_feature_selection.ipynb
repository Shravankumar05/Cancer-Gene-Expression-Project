```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x_train = pd.read_csv('../data/processed/x_train.csv', index_col=0)
y_train = pd.read_csv('../data/processed/y_train.csv')['label'].values
x_test = pd.read_csv('../data/processed/x_test.csv', index_col=0)
y_test = pd.read_csv('../data/processed/y_test.csv')['label'].values

import joblib
le = joblib.load('../data/processed/label_encoder.joblib')
print(f"X_train shape: {x_train.shape}")
print(f"X_test shape:  {x_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape:  {y_test.shape}")
print(f"Classes: {le.classes_}")
```

```python
var_scaled = x_train.var()
print(var_scaled.describe())
```

Every quartile is equal since we had normalised the gene variance to variance ~1 across the samples (not exactly 1 as explained earlier). With genes at variance 0 that are completely dead across all the samples.

```python
print(f'genes with no variance: {(var_scaled == 0).sum()}')
```

There are 276 dead genes with no variance ready to be removed.

```python
var_scaled = x_train.var()
kept_genes = x_train.columns[var_scaled > 0]

x_train = x_train[kept_genes]
x_test = x_test[kept_genes]

print(len(kept_genes))
```
